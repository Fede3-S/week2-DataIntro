## 1. Research Question

- **Main research question:** 
Can superconductive materials be clustered into chemically or structurally distinct groups using unsupervised ML?
- **Why is this question interesting or important?** 
Understanding superconductivity is one of the central challenges in modern condensed matter physics. While high-temperature superconductors hold transformative potential for energy transmission, quantum computing, and medical technology, discovering new superconductive materials has historically relied on trial-and-error experimentation or incremental tweaks to known chemical families.
- **Who might use or care about the answer?** 
Materials Scientists & Computational Chemists: To discover unexplored chemical spaces and establish data-driven design rules for new superconductors.
Condensed Matter Physicists: To test whether machine-learned clusters align with theoretical models or reveal anomalous physical behaviors.
ML & Materials Informatics Engineers: To build better prediction models by partitioning complex datasets into distinct, specialized sub-models.
Energy & Tech R&D Teams: To identify viable candidates for commercial applications like quantum computing, fusion reactors, and power grids.

Which chemical properties drive the primary cluster boundaries?

How do density-based methods compare to centroid-based methods in separating complex material groups?

### 2. Why was it originally collected?

Why was the dataset originally collected?
The dataset was collected to compile information about known superconducting materials and their properties, particularly their critical temperature. The materials have also been featurised using properties derived from their chemical compositions, such as electronegativity, atomic weight, valence electrons, ionic character and atomic size. This makes the dataset useful for studying relationships between material composition and superconducting behaviour, as well as for applying machine-learning methods to superconductors.

### 3. Who is represented?

The dataset contains 16,521 superconducting materials belonging to several different superconductor types. The largest groups are:
- Other: 6,733
- Cuprates: 5,941
- Iron-based: 1,800
- Bismuthates: 1,553
- Borocarbides: 349
- Organic superconductors: 75
- Elemental superconductors: 64
- Hydrogen-rich superconductors: 6
Therefore, the dataset is dominated by the Other and Cuprate categories, while some superconductor families have very little representation.

### 4. Who might be missing?

Some superconductor types are likely to be underrepresented rather than completely absent. For example, there are only six hydrogen-rich superconductors and relatively few organic and elemental superconductors. There may also be superconducting materials or proposed candidates that are not included in the source data.
This imbalance is important for clustering. Large groups such as cuprates may have a much stronger influence on the patterns found by an unsupervised algorithm, while very small families may not form identifiable clusters at all.

### 5. What personal or sensitive information does it contain, explicitly or implicitly?

The dataset contains no personal information. Each observation represents a material rather than a person. The columns describe chemical composition and material properties, such as, electronegativity, atomic properties and valence-electron characteristics.
There also is no sensitive personal information that could be inferred from these variables. Therefore, privacy and personal-data concerns are minimal compared with datasets involving human subjects.

### 6. What could your analysis reveal that was not obvious in the original data?

Clustering could reveal whether superconductors naturally separate into groups based only on their physical and chemical features, rather than their existing category labels. These clusters could then be compared with known superconductor types to determine whether the known families occupy distinct regions of feature space.
Further analysis could reveal clusters that correspond to known families, subdivisions within existing categories, similarities between supposedly different types, or unusual materials that do not fit their expected group.

### 7. Who could benefit from your findings?

The findings could benefit superconductivity and materials-science researchers, particularly those investigating relationships between chemical composition, material properties and superconducting behaviour. Identifying natural groupings could help researchers understand similarities between known superconductors and potentially identify unusual or interesting materials for further investigation.
Machine-learning researchers working in materials science could also benefit because the analysis can show which features are most useful for distinguishing superconducting material types and whether dimensionality-reduction and clustering methods can reproduce established physical classifications.

### 8. Who could be harmed or misrepresented?

While material datasets carry no direct risk to human subjects, misrepresentation can still cause indirect harm:

Experimental Researchers: Flawed clusters or false patterns could lead researchers to waste lab time and funding pursuing unviable candidate materials.

Underrepresented Material Families: Small groups (like hydrogen-rich or organic superconductors) risk being dismissed as statistical outliers or noise simply due to sample size.

Scientific Literature: Presenting purely algorithmic groupings as true physical laws without domain validation risks misrepresenting established physics.
